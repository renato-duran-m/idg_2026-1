## 1. Introducción

La toma de decisiones basada en datos espaciales es hoy un pilar fundamental para la modernización del retail y el comercio minorista. La motivación central de este estudio es aplicar técnicas de microsimulación y agrupamiento territorial para resolver un desafío comercial real: el proyecto de expansión de la Botillería 7LIRAS.

El objetivo primordial para 7LIRAS es transitar desde un modelo de ventas puramente físico y tradicional hacia la implementación de un canal de ventas digital, con el fin de ampliar su alcance geográfico más allá de su ubicación actual e incrementar su rentabilidad anual. Para que esta expansión sea exitosa y eficiente en el uso de capital, es indispensable comprender cómo se distribuyen demográfica y económicamente los potenciales consumidores en el Sector Sur-Poniente del Gran Santiago (Maipú, Cerrillos, Estación Central y Pudahuel).

### 1.1 Marco Conceptual: E-commerce y Dark Store

Para materializar este proyecto, la estrategia de expansión se fundamenta en la implementación de una dark store. En la logística urbana moderna, las dark stores se han consolidado como nodos de distribución esenciales para el comercio electrónico, ya que al ubicarse estratégicamente en zonas de alta densidad, permiten minimizar la distancia de la 'última milla', reduciendo significativamente los costos operativos y los tiempos de despacho al cliente final (Buldeo Rai et al., 2019).

* **E-commerce (Comercio Electrónico):** Consiste en la digitalización del canal de ventas, permitiendo a los clientes explorar el catálogo, realizar compras y procesar pagos a través de plataformas digitales (sitios web o aplicaciones de delivery). Esto elimina la barrera geográfica de la tienda física tradicional.

* **Dark Store (Tienda Oscura o Bodega Oculta):** Es un centro de distribución o micro-almacén exclusivo para operaciones de comercio electrónico. A diferencia de una tienda física, una dark store no está abierta al público general ni cuenta con vitrinas o cajeros; su distribución interna está 100% optimizada para la preparación rápida de pedidos (picking y packing) y su posterior despacho mediante flotas de última milla. Identificar el territorio correcto para ubicar una dark store es vital para minimizar los tiempos y costos de entrega.

### 1.2 Zona de estudio y Justificación
Se analizará el **Sector Sur-Poniente del Gran Santiago**, compuesto por las comunas de **Maipú, Cerrillos, Estación Central y Pudahuel**. 
La elección de este eje territorial resulta estratégica para identificar patrones demográficos y económicos útiles en la implementación de un canal de ventas e-commerce para la Botillería 7LIRAS. Identificar estas tipologías permite planificar la cobertura de última milla, evaluar la ubicación de *dark stores* y ampliar el alcance geográfico con rentabilidad en un sector de alta conectividad.

### 1.3 Variables Seleccionadas
*   **Variable a Microsimular (CASEN): `ytotcor` (Ingreso total del hogar).** La primera variable estratégica seleccionada es el Ingreso Total del Hogar. La literatura empírica sobre comercio electrónico demuestra que el nivel de ingresos es uno de los principales predictores del comportamiento de compra digital, ya que determina no solo la frecuencia de las transacciones, sino también el valor del ticket promedio y la disposición a pagar por servicios de conveniencia como el delivery (Clemes et al., 2014). Conocer el presupuesto de los hogares permite estructurar el mix de productos adecuado para cada sector.
*   **Variable Censo 1: Porcentaje de jóvenes adultos (18 a 44 años).** La segunda variable corresponde al Porcentaje de Jóvenes Adultos. Estudios sobre la demografía del consumidor digital indican que las generaciones Millennial y Gen Z son los principales adoptantes del quick commerce (comercio rápido) y aplicaciones de reparto. Este grupo etario valora por sobre todo la inmediatez y la conveniencia tecnológica, convirtiéndolos en el segmento objetivo natural para la venta de licores a domicilio a través de plataformas digitales (Vasić et al., 2019).
*   **Variable Censo 2: Porcentaje de viviendas en arriendo.** Finalmente, se seleccionó el Porcentaje de Viviendas en Arriendo. En estudios de geografía económica y logística urbana, las zonas con altas tasas de arriendo se correlacionan directamente con una mayor densificación habitacional (edificación en altura) y poblaciones transitorias con estilos de vida acelerados. Estas características espaciales son fundamentales para el e-commerce, ya que garantizan una alta concentración de demanda por metro cuadrado, lo cual maximiza la eficiencia y rentabilidad de las rutas de despacho (Morganti et al., 2014).

### 1.4 Cálculo de indicadores

* **Estimación del Ingreso Total:** Dado que el Censo no recopila información financiera, el ingreso se estimó utilizando el algoritmo de Microsimulación Espacial IPF (Ajuste Proporcional Iterativo). Este algoritmo "clonó" perfiles de la CASEN para rellenar las zonas censales, ajustándose iterativamente hasta que la edad, sexo, educación y tenencia de la vivienda coincidieran con los datos censales reales, obteniendo finalmente la mediana del ingreso simulado por zona.

* **Cálculo del % de Jóvenes Adultos:** Se sumó la población censada en los rangos de 18 a 24 años y de 25 a 44 años. Este subtotal se dividió por la población total de la zona censal respectiva (n_per) y se multiplicó por 100.

* **Cálculo del % de Arriendo:** Se sumaron los hogares bajo régimen de tenencia arrendada (con y sin contrato), se dividió dicho valor por el total de hogares de la zona censal (n_hog) y se multiplicó por 100.

### 1.5 Objetivo

Finalmente, estos tres indicadores fueron estandarizados numéricamente y procesados mediante el algoritmo de agrupamiento K-Means, con el propósito de identificar tipologías territoriales que guíen la instalación de las futuras dark stores de 7LIRAS y proporcione información para adoptar ofertas personalizadas para cada zona.

### 2.1.1 Librería

```python
!pip install ipfn

import pandas as pd
import geopandas as gpd
import matplotlib.pyplot as plt
import matplotlib.patheffects as pe
import seaborn as sns
import numpy as np
import pyreadr
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from ipfn import ipfn
import contextlib
import io
import warnings
```

### 2.1.2 Conexión Base de Datos

```python
sns.set_theme(style="whitegrid")
SEED = 123

# Rutas a los archivos
RUTA_CENSO_ZONAL = '../Parquet/Cartografía_censo2024_R13_Zonal.parquet'
RUTA_CENSO_COMUNAL = '../Parquet/Cartografía_censo2024_R13_Comunal.parquet'
RUTA_CASEN = '../Parquet/casen_rm.rds'
```

### 2.1.3 Datos censo y límites comunales

```python
# 1. Carga de datos Censo y Filtro Espacial
censo_zonal = gpd.read_parquet(RUTA_CENSO_ZONAL).rename_geometry("geometry")
comunas_estudio = ["MAIPÚ", "CERRILLOS", "ESTACIÓN CENTRAL", "PUDAHUEL"]
es_zona_estudio = censo_zonal["COMUNA"].isin(comunas_estudio)
censo_zonal = censo_zonal[es_zona_estudio].reset_index(drop=True)

censo_comunal = gpd.read_parquet(RUTA_CENSO_COMUNAL).rename_geometry("geometry")
censo_comunal = censo_comunal[censo_comunal["COMUNA"].isin(comunas_estudio)].reset_index(drop=True)

# 2. Carga Casen
casen_raw = pyreadr.read_r(RUTA_CASEN)
casen_raw = casen_raw[list(casen_raw.keys())[0]]
```

### 2.2 Pre-procesamiento CASEN

```python
# 1. DEFINICIÓN DE VARIABLES
# Variables base de la CASEN que no se alteran y se usarán como restricciones 
VARS_FIJAS = ["estrato","esc","e6a","edad","sexo", "v13","v13_propia","v13_arrendada","v13_cedida"]

# Variable objetivo a microsimular (Ingreso total del hogar)
VARIABLE_A_MICROSIMULAR = "ytotcor" 
FUNCION_AGREGACION = np.median # Usaremos la mediana para evitar sesgos por valores atípicos (outliers)

# Filtramos la CASEN solo con las variables necesarias
casen = casen_raw[VARS_FIJAS + [VARIABLE_A_MICROSIMULAR]].copy()
del casen_raw

# Extraemos la comuna desde la variable estrato y aseguramos que sea formato entero
casen["comuna"] = (casen["estrato"]//100).astype(int)
casen = casen.drop(columns="estrato")

# 2. PREPARACIÓN DE CONSTRAINTS (CENSO)
ZONA_ID = "OBJECTID"

# Diccionarios de homologación entre variables CASEN y columnas del Censo
COLS_SEXO = {"hombre": "n_hombres", "mujer": "n_mujeres"}
COLS_EDAD = {"edad_0_5": "n_edad_0_5", "edad_6_13": "n_edad_6_13", "edad_14_17": "n_edad_14_17", 
             "edad_18_24": "n_edad_18_24", "edad_25_44": "n_edad_25_44", "edad_45_59": "n_edad_45_59", "edad_60_mas": "n_edad_60_mas"}
COLS_EDUC = ["educ_nunca", "educ_primaria", "educ_secundaria", "educ_terciaria"]
COLS_TENENCIA = {"ten_propia_pagada": "n_tenencia_propia_pagada", "ten_propia_pagandose": "n_tenencia_propia_pagandose",
                 "ten_arrendada_contrato": "n_tenencia_arrendada_contrato", "ten_arrendada_sin_contrato": "n_tenencia_arrendada_sin_contrato",
                 "ten_cedida_trabajo": "n_tenencia_cedida_trabajo", "ten_cedida_familiar": "n_tenencia_cedida_familiar", "ten_otro": "n_tenencia_otro"}

# Inicializamos el DataFrame de restricciones zonales
constraints = censo_zonal[[ZONA_ID, "CUT", "COMUNA", "n_per", "n_hog"]].copy()

# Forzamos a que el CUT del Censo sea entero exacto para que el cruce posterior no falle
constraints["CUT"] = constraints["CUT"].astype(int)

# Agregamos los conteos de sexo y edad directamente
for nombre,col in {**COLS_SEXO, **COLS_EDAD}.items(): 
    constraints[nombre] = censo_zonal[col]

# Agrupamos las categorías de educación del Censo en 4 niveles para cruzar con CASEN
constraints["educ_nunca"] = censo_zonal["n_cine_nunca_curso_primera_infancia"]
constraints["educ_primaria"] = censo_zonal["n_cine_primaria"] + censo_zonal["n_cine_especial_diferencial"]
constraints["educ_secundaria"] = censo_zonal["n_cine_secundaria"]
constraints["educ_terciaria"] = censo_zonal["n_cine_terciaria_maestria_doctorado"]

# La tenencia en el Censo está a nivel de hogar, la extrapolamos a nivel de personas
personas_por_hogar = np.where(censo_zonal["n_hog"] > 0, censo_zonal["n_per"] / censo_zonal["n_hog"], 0)
for nombre, col in COLS_TENENCIA.items(): 
    constraints[nombre] = censo_zonal[col] * personas_por_hogar

CATEGORIAS = {"sexo_cat": list(COLS_SEXO.keys()), "edad_cat": list(COLS_EDAD.keys()), "educ_cat": COLS_EDUC, "tenencia_cat": list(COLS_TENENCIA.keys())}
constraints = constraints.set_index(ZONA_ID)

# 3. RECODIFICACIÓN CASEN 
# Imputamos los datos faltantes de escolaridad (esc) usando una Regresión Lineal sobre el último curso aprobado (e6a)
falta_esc = casen["esc"].isna()
modelo_esc = LinearRegression().fit(casen.loc[~falta_esc, ["e6a"]], casen.loc[~falta_esc, "esc"])
casen.loc[falta_esc, "esc"] = np.clip(np.round(modelo_esc.predict(casen.loc[falta_esc, ["e6a"]])), 0, 20)

# Discretización (pd.cut) para categorizar la CASEN en los mismos tramos del Censo
casen["sexo_cat"] = casen["sexo"].map({1: "hombre", 2: "mujer"})
casen["edad_cat"] = pd.cut(casen["edad"], bins=[0, 6, 14, 18, 25, 45, 60, np.inf], labels=list(COLS_EDAD.keys()), right=False, include_lowest=True).astype(str)
casen["educ_cat"] = pd.cut(casen["esc"], bins=[-1, 0, 8, 12, np.inf], labels=COLS_EDUC).astype(str)

# Recodificación de Tenencia de la vivienda usando np.select para múltiples condiciones
condiciones = [(casen["v13"]==1)&(casen["v13_propia"]==1), (casen["v13"]==1)&(casen["v13_propia"]==2), (casen["v13"]==2)&(casen["v13_arrendada"]==5), (casen["v13"]==2)&(casen["v13_arrendada"]==6), (casen["v13"]==3)&(casen["v13_cedida"]==7), (casen["v13"]==3)&(casen["v13_cedida"]==8)]
casen["tenencia_cat"] = np.select(condiciones, ["ten_propia_pagada", "ten_propia_pagandose", "ten_arrendada_contrato", "ten_arrendada_sin_contrato", "ten_cedida_trabajo", "ten_cedida_familiar"], default="ten_otro")

# Excluimos de la muestra a quienes no tienen dato en la variable a microsimular
casen = casen[~casen[VARIABLE_A_MICROSIMULAR].isna()].reset_index(drop=True)

# Finalmente, aseguramos dejar solo a los donantes que pertenecen a las comunas de estudio
casen = casen[casen["comuna"].isin(constraints["CUT"].unique())].reset_index(drop=True)
```

### 2.3 Algoritmo de microsimulación

```python
# 1. FUNCIONES PRINCIPALES DE MICROSIMULACIÓN 

def pesos_por_persona(seed, codigos, celda_a_personas):
    """
    Función que aplica el algoritmo Iterative Proportional Fitting (IPFN).
    Ajusta iterativamente los pesos de la muestra CASEN (seed) para que 
    sus totales coincidan con los márgenes del Censo en esa zona (objetivos).
    """
    # Se le indica al algoritmo sobre qué dimensiones aplicar cada restricción
    dims = [[i] for i in range(len(codigos["objetivos"]))]
    
    # Se ejecuta IPFN con una tasa de convergencia de 1e-6 y máximo 50 iteraciones
    IPF = ipfn.ipfn(seed.copy(), codigos["objetivos"], dims, convergence_rate=1e-6, max_iteration=50)
    
    # Devuelve el peso fraccionario INDIVIDUAL de cada donante
    return IPF.iteration()[codigos["indices"]] / celda_a_personas

def integerise_trs(pesos, rng):
    """
    Función Truncate-Replicate-Sample (TRS) para integerización.
    Transforma los pesos fraccionarios generados por IPF en una población entera.
    """
    # 1. Truncate (Trunco): Se toma la parte entera del peso
    parte_entera = np.floor(pesos).astype(int)
    
    # 2. Sample (Muestra): Se guarda la parte decimal para sortearla posteriormente
    parte_fraccionaria = pesos - parte_entera
    
    idx = np.arange(len(pesos))
    replicados = np.repeat(idx, parte_entera) # Replicate
    
    # Calculamos cuántas personas faltan sortear para calzar con la población zonal
    n_faltante = int(round(pesos.sum())) - parte_entera.sum()
    
    if n_faltante > 0 and parte_fraccionaria.sum() > 0:
        probs = parte_fraccionaria / parte_fraccionaria.sum()
        # Sorteamos a los individuos faltantes según su probabilidad
        personas_extra = rng.choice(idx, size=n_faltante, replace=True, p=probs)
        replicados = np.concatenate([replicados, personas_extra])
        
    return replicados
```

### 2.4 Ejecución del algoritmo

```python
# 1. EJECUCIÓN DEL BUCLE DE MICROSIMULACIÓN ESPACIAL 
rng = np.random.default_rng(SEED) 
VARS_CONSTRAINT = list(CATEGORIAS.keys())

resultados_zona = []
zonas_sin_donantes = []

buffer_salida = io.StringIO()
with contextlib.redirect_stdout(buffer_salida), warnings.catch_warnings():
    warnings.simplefilter("ignore", RuntimeWarning)
    
    # Itración 1: Agrupación comuna
    for comuna_cut, cons_comuna in constraints.groupby("CUT"):
        muestra_comuna = casen[casen["comuna"] == comuna_cut]
        
        # Si no hay donantes en una comuna, se toma toda la muestra
        if len(muestra_comuna) == 0: 
            muestra_comuna = casen.copy()
            zonas_sin_donantes.extend(cons_comuna.index.tolist())

        # Construimos la "matriz semilla" inicial
        indices = {var: pd.Categorical(muestra_comuna[var], categories=CATEGORIAS[var]).codes for var in VARS_CONSTRAINT}
        forma = tuple(len(CATEGORIAS[var]) for var in VARS_CONSTRAINT)
        seed_tabla = np.zeros(forma)
        indices_tupla = tuple(indices[var] for var in VARS_CONSTRAINT)
        np.add.at(seed_tabla, indices_tupla, 1)
        
        # TRUCO DE LAPLACE: Sumamos 0.001 para evitar "Ceros Estructurales"
        # Esto previene que el algoritmo colapse al dividir por cero en combinaciones vacías de la CASEN
        seed_tabla += 0.001
        
        celda_a_personas = seed_tabla[indices_tupla] 
        valores_objetivo = muestra_comuna[VARIABLE_A_MICROSIMULAR].to_numpy()

        # Iteración 2: Por zona censal
        for zona_id, fila in cons_comuna.iterrows():
            if fila["n_per"] == 0: continue
            
            objetivos = [np.array([fila[cat] for cat in CATEGORIAS[var]], dtype=float) for var in VARS_CONSTRAINT]
            
            # Bloque try-except por robustez ante zonas extremadamente atípicas
            try:
                pesos = pesos_por_persona(seed_tabla, {"objetivos": objetivos, "indices": indices_tupla}, celda_a_personas)
                sinteticos_idx = integerise_trs(pesos, rng)
                valores_sinteticos = valores_objetivo[sinteticos_idx]
                
                # Agregamos la Mediana del ingreso total del hogar simulado
                resultados_zona.append({
                    "zona_id": zona_id, 
                    "CUT": comuna_cut,
                    "n_sintetico": len(sinteticos_idx),
                    "n_censo": fila["n_per"],
                    "valor_zona": FUNCION_AGREGACION(valores_sinteticos) if len(valores_sinteticos) else np.nan
                })
            except Exception:
                continue

# 2. Resultados espaciales
resultados_zona = pd.DataFrame(resultados_zona).set_index("zona_id")
```

### 2.4.1 Mapa variable microsimulada

```python
if "valor_zona" in censo_zonal.columns:
    zonas_mapa = censo_zonal.set_index(ZONA_ID).copy()
else:
    zonas_mapa = censo_zonal.set_index(ZONA_ID).join(resultados_zona[["valor_zona"]])
    censo_zonal = censo_zonal.set_index(ZONA_ID).join(resultados_zona[["valor_zona"]]).reset_index()

# 1. MAPA DE LA VARIABLE MICROSIMULADA Y ESTADÍSTICAS
fig, ax = plt.subplots(figsize=(10, 10))
zonas_mapa.plot(
    column="valor_zona",
    cmap="viridis",
    linewidth=0.05,
    edgecolor="grey",
    legend=True,
    missing_kwds={"color": "lightgrey", "label": "sin dato"},
    ax=ax,
)
ax.set_title(f"{FUNCION_AGREGACION.__name__}({VARIABLE_A_MICROSIMULAR}) microsimulado por zona censal\nSector Sur-Poniente")
ax.set_axis_off()
plt.show()

# Estadísticas
print(f"Zonas simuladas: {len(resultados_zona)}")
print(f"Zonas sin comuna con donantes CASEN: {len(zonas_sin_donantes)}")
display(resultados_zona.head())
```

### 2.4.2 Interpretación Espacial de la Variable Microsimulada: Ingreso Total del Hogar (ytotcor)

La visualización de la mediana del Ingreso Total del Hogar microsimulado sobre la cartografía censal revela un patrón de distribución económica profundamente heterogéneo dentro del Sector Sur-Poniente. Esta primera capa analítica ofrece directrices fundamentales para la estructuración comercial del nuevo canal digital.

**Patrones Espaciales Observados:**
* **Concentración de Altos Ingresos (Zonas Amarillas):** Se evidencia alto poder adquisitivo muy focalizado, con medianas simuladas que superan los \$350.000, ubicado predominantemente en el cuadrante nororiente del polígono de estudio (límite entre Estación Central y Pudahuel).
*  **Predominio de Ingresos Medios (Zonas Verdes):** La gran masa territorial del área de estudio, especialmente extensa hacia el centro y sur (ejes principales de Maipú y Cerrillos), fluctúa en un rango de ingresos medios consolidados (entre \$150.000 y \$250.000).
*   **Focos de Vulnerabilidad (Zonas Moradas/Azul Oscuro):** Se detectan zonas altamente localizadas donde la mediana de ingresos cae abruptamente (por debajo de los \$100.000), representando sectores con restricciones estructurales para el consumo de bienes no esenciales.

**Implicancias Estratégicas para la Expansión Comercial:**
Desde la perspectiva operativa del comercio electrónico, este mapa unidimensional permite establecer las primeras hipótesis para la estrategia de catálogo y posicionamiento:
*   **Segmentación del Catálogo (Margen vs. Volumen):** El rango amarillo nororiente constituye el nicho geográfico óptimo para dirigir pautas publicitarias digitales enfocadas en productos de alto margen (licores premium, destilados de especialidad y formatos de regalo). En contraste, la vasta extensión de las zonas medias sugiere que la oferta del e-commerce debe sostenerse mediante un catálogo de alta rotación, ofertas transaccionales por volumen y packs orientados al consumo de fin de semana.
*   **Logística y Ubicación de la Dark Store:** El mapa advierte que instalar la bodega exclusivamente dentro del foco de mayores ingresos podría ser ineficiente debido a los probables sobrecostos de arriendo comercial (suelo más caro). La lectura espacial sugiere que la ubicación ideal de una dark store debería ser un punto pivotante dentro de las zonas de ingresos medios, asegurando arriendos logísticos más bajos, pero garantizando radios de despacho ultrarrápidos hacia el clúster amarillo de alto consumo.

**Justificación para el Modelo Multidimensional (Clustering):**
Si bien este mapa de poder adquisitivo es un piso analítico vital, resulta insuficiente por sí solo para la toma de decisiones en canales digitales. En el comercio de última milla, una zona de altos ingresos habitada por familias mayores y propietarias suele registrar menor adopción de delivery que una zona de ingresos medios habitada por jóvenes profesionales y arrendatarios transitorios. Esta limitación es la que justifica metodológicamente la necesidad de integrar las tasas demográficas (% Jóvenes y % Arriendo) para ejecutar el algoritmo de agrupamiento multivariable.

### 2.5 Escalamiento y método del codo

```python
# 1. FILTRADO DE ZONAS VÁLIDAS 
# Mantenemos solo zonas con población, hogares y un valor de ingreso microsimulado válido
usable = (censo_zonal['n_per'] > 0) & (censo_zonal['n_hog'] > 0) & (censo_zonal['valor_zona'].notna())
censo_zonal = censo_zonal[usable].reset_index(drop=True)

# 2. CONSTRUCCIÓN DE INDICADORES TERRITORIALES (CENSO) 
# Porcentaje de Jóvenes Adultos (18 a 44 años) 
censo_zonal['pct_jovenes'] = ((censo_zonal['n_edad_18_24'] + censo_zonal['n_edad_25_44']) / censo_zonal['n_per']) * 100

# Porcentaje de Viviendas en Arriendo 
censo_zonal['pct_arriendo'] = ((censo_zonal['n_tenencia_arrendada_contrato'] + censo_zonal['n_tenencia_arrendada_sin_contrato']) / censo_zonal['n_hog']) * 100

# Seleccionamos las variables finales
indicadores = ['valor_zona', 'pct_jovenes', 'pct_arriendo']
X = censo_zonal[indicadores]

# 3. ESCALAMIENTO DE VARIABLES 
# K-Means agrupa usando distancia Euclidiana. Como el ingreso está en cientos de miles 
# y los porcentajes de 0 a 100, aplicamos StandardScaler (media 0, varianza 1) 
# para evitar que la magnitud del ingreso domine la formación de los clusters.
escalador = StandardScaler()
X_escalado = escalador.fit_transform(X)
```

### 2.5.1 Busqueda K óptimo y correlaciones

```python
# 1. BÚSQUEDA DE K ÓPTIMO (MÉTODO DEL CODO)
# Calculamos la inercia (WSS) probando desde 1 hasta 10 clusters (K)
wss = []
valores_k = range(1, 11)
for k in valores_k:
    modelo = KMeans(n_clusters=k, n_init=25, random_state=SEED)
    modelo.fit(X_escalado)
    wss.append(modelo.inertia_)

plt.figure(figsize=(7, 5))
plt.plot(valores_k, wss, marker="o", color="#1f77b4", linewidth=2)
plt.title("Método del Codo (Ingreso Hogar, % Jóvenes, % Arriendo)", fontsize=12, fontweight='bold')
plt.xlabel("Número de clusters (k)", fontsize=10)
plt.ylabel("Inercia (WSS)", fontsize=10)
plt.xticks(list(valores_k))
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()

# 2. CORRELACIONES 
print("Correlación Ingreso vs % Jóvenes:")
print(censo_zonal["valor_zona"].corr(censo_zonal["pct_jovenes"]))

print("\nCorrelación Ingreso vs % Arriendo:")
print(censo_zonal["valor_zona"].corr(censo_zonal["pct_arriendo"]))

print("\nCorrelación % Jóvenes vs % Arriendo:")
print(censo_zonal["pct_jovenes"].corr(censo_zonal["pct_arriendo"]))
```

### 2.5.2 Determinación del Número Óptimo de Clusters (Método del Codo)

Para ejecutar el algoritmo K-Means, es un requisito metodológico definir la cantidad de grupos (K) en los que se dividirá el territorio. Para evitar una selección arbitraria que pudiera sesgar la estrategia de la **Botillería 7LIRAS**, se utilizó el Método del Codo (*Elbow Method*), evaluando la Inercia o Suma de Cuadrados Intra-Cluster (WSS) para iteraciones del 1 al 10.

Al analizar el gráfico resultante ("Método del Codo"), la justificación para seleccionar **K=4** se sostiene en dos pilares fundamentales: uno matemático y otro comercial.

**Justificación Matemática (El Punto de Inflexión):**
El objetivo del algoritmo es minimizar la distancia interna de los elementos dentro de cada cluster (hacerlos lo más compactos y similares posible) sin crear una cantidad excesiva de grupos. 
*   En la curva generada, se observa una caída drástica de la inercia desde k=1 hasta k=3, lo que indica que el modelo está aprendiendo y separando grandes diferencias estructurales en los datos.
*   Al llegar a **k=4**, la inercia se ubica levemente por encima del valor 200. 
*   A partir de **k=5 en adelante**, la curva sufre un quiebre evidente (formando el "codo" geométrico) y tiende a aplanarse. Esto significa que agregar un quinto, sexto o séptimo cluster aporta reducciones marginales de varianza muy bajas. Matemáticamente, k=4 representa el punto de equilibrio óptimo entre la máxima compresión de datos y la parsimonia del modelo.

**Justificación Comercial y Estratégica:**
Desde la perspectiva de la gestión del e-commerce y la logística de distribución, el número de segmentos debe ser operable.
*   **Si se eligiera un K menor (ej. K=2):** Se caería en una sobresimplificación del mercado (por ejemplo, dividiendo el territorio únicamente en "ingresos altos" vs "ingresos bajos"), perdiendo toda la riqueza que aportan las variables de juventud y arriendo.
*   **Si se eligiera un K mayor (ej. K=7):** Se generaría una hiper-segmentación territorial. Para una botillería en fase de expansión digital, diseñar siete catálogos distintos o siete estrategias de ruteo de última milla resulta ineficiente, encareciendo los costos de marketing y complicando la operación de la *dark store*.
*   **El balance de K=4:** Cuatro tipologías territoriales permiten estructurar una matriz de decisiones clara y diferenciada. Facilita la creación de campañas de marketing digital muy bien dirigidas (ej. un segmento premium, un segmento joven/dinámico, un segmento familiar/tradicional, etc.) manteniendo los costos logísticos centralizados y controlados.

### 2.5.3 Análisis de Correlación de Variables

Previo a la ejecución del algoritmo de agrupamiento multivariable, se calculó la correlación lineal de Pearson entre los tres indicadores territoriales para comprender cómo interactúan espacialmente. Los resultados obtenidos confirman las hipótesis demográficas y justifican la pertinencia del modelo K-Means para la estrategia comercial de expansión de la Botillería 7LIRAS.

**1. Correlación % Jóvenes vs % Arriendo (r = 0.673):**
Esta es la relación positiva más fuerte entre las variables observadas. Indica de manera robusta que, a medida que aumenta la proporción de viviendas en arriendo en una zona censal, también lo hace significativamente la concentración de jóvenes adultos. Desde la perspectiva de negocio, esto valida teóricamente el perfil del consumidor digital: las zonas de alta densificación y rotación habitacional están ocupadas precisamente por la generación con mayor adopción tecnológica y uso intensivo de aplicaciones de delivery.

**2. Correlación Ingreso vs % Arriendo (r = 0.634):**
Se observa una correlación positiva de fuerza moderada-alta, lo cual constituye un hallazgo geográfico sumamente revelador para el Sector Sur-Poniente. Este indicador sugiere que el mercado de arriendo en esta zona no está asociado a vulnerabilidad económica, sino que, por el contrario, las zonas con mayor concentración de arriendos tienden a registrar mayores ingresos. Este fenómeno es característico del desarrollo inmobiliario en ejes de alta conectividad (como los corredores de metro y principales avenidas), atrayendo a profesionales jóvenes con alta liquidez y disposición a pagar por servicios de conveniencia rápida.

**3. Correlación Ingreso vs % Jóvenes (r = 0.279):**
Esta correlación positiva es débil. Significa que, aunque existe una ligera tendencia, la sola presencia de población joven no garantiza un alto nivel de ingresos en la zona. La juventud se encuentra distribuida de manera transversal a lo largo de distintos estratos socioeconómicos dentro de las comunas evaluadas.

**Implicancia Metodológica y Estratégica:**
Estos índices justifican matemáticamente la aplicación del modelo multivariable. Si las tres variables tuvieran correlaciones perfectas (cercanas a 1), bastaría con mapear el ingreso para tomar decisiones. Sin embargo, la debilidad del cruce entre "Ingreso" y "Juventud" demuestra que estas dimensiones aportan información independiente y complementaria. El algoritmo K-Means tomará estas variables entrelazadas para encontrar el rango de valor exacto; aislando, por ejemplo, los clústeres de jóvenes arrendatarios de altos ingresos (el target óptimo para posicionar publicidad de licores premium y ubicar la dark store) del resto del territorio.

### 2.6 K-Means

```python
# Fijamos el K óptimo basándonos en el método del codo
K = 4 

# Ejecutamos el algoritmo K-Means final y almacenamos las etiquetas
kmeans = KMeans(n_clusters=K, n_init=25, random_state=SEED)
censo_zonal["cluster"] = kmeans.fit_predict(X_escalado).astype(str)
```

### 2.7 Cálculo y mapa del Índice de Shannon

```python
# 1. CÁLCULO DE HETEROGENEIDAD (ÍNDICE DE SHANNON) 
# La entropía de Shannon (H) penaliza la concentración. 
# Si una comuna tiene zonas repartidas equitativamente en todos los clusters, su índice será alto.
def calcular_shannon(serie_clusters):
    if len(serie_clusters) == 0: return 0
    # Obtenemos la proporción (p_i) de cada cluster
    proporciones = serie_clusters.value_counts(normalize=True)
    # H = -sum(p_i * ln(p_i))
    return -np.sum(proporciones * np.log(proporciones))

# Aplicamos la función agrupando las zonas a nivel de comuna
shannon_df = censo_zonal.groupby("COMUNA")["cluster"].apply(calcular_shannon).reset_index(name="indice_shannon")

# Join espacial para integrar el índice a la cartografía comunal
mapa_comunas = censo_comunal.merge(shannon_df, on="COMUNA")
```

### 2.7.1 Mapa de entropía

```python
# 1. VISUALIZACIÓN DEL MAPA DE ENTROPÍA 
fig, ax = plt.subplots(figsize=(10, 10))

mapa_comunas.plot(column="indice_shannon", cmap="magma_r", linewidth=0.8, edgecolor="black", legend=True, legend_kwds={'label': "Índice de Shannon"}, ax=ax)

for _, fila in mapa_comunas.iterrows():
    punto = fila.geometry.representative_point()
    ax.annotate(f"{fila['COMUNA'].title()}\n({fila['indice_shannon']:.2f})", xy=(punto.x, punto.y), ha="center", va="center", fontsize=8, color="black", path_effects=[pe.withStroke(linewidth=2, foreground="white")])

ax.set_title("Variabilidad Intra-comunal de Tipologías Territoriales (Índice de Shannon)")
ax.set_axis_off()
ax.set_xlim(xmin - margen_x, xmax + margen_x)
ax.set_ylim(ymin - margen_y, ymax + margen_y)
plt.show()

print("Índices de Shannon por comuna:")
display(mapa_comunas[['COMUNA', 'indice_shannon']].sort_values('indice_shannon', ascending=False))
```

### 2.7.2 Variabilidad Intra-comunal: Índice de Shannon

Para cuantificar matemáticamente la heterogeneidad visualizada en el mapa de *clusters*, se calculó el Índice de Shannon a nivel comunal. En este contexto de inteligencia territorial, este indicador mide la "diversidad de tipologías": un valor alto indica que dentro de una misma comuna conviven múltiples perfiles socioeconómicos en proporciones similares, mientras que un valor bajo señala que la comuna es altamente uniforme y está dominada por uno o dos perfiles mayoritarios.

La proyección cartográfica de este índice arroja dos realidades territoriales opuestas dentro del Sector Sur-Poniente:

**Alta Fragmentación y Diversidad (Estación Central):**
*   El mapa destaca visualmente a **Estación Central** con una tonalidad negra, alcanzando un Índice de Shannon de **1.28**, el valor más alto de la muestra por un margen considerable. 
*   **Interpretación analítica:** Esta comuna funciona como un mosaico demográfico. Su territorio no obedece a un patrón único, sino que está altamente fragmentado. Como se adelantaba en la cartografía anterior, en distancias muy cortas conviven simultáneamente perfiles radicales, desde sectores de muy altos ingresos y arriendo, hasta zonas de vulnerabilidad.

**Homogeneidad Territorial (Maipú, Pudahuel y Cerrillos):**
*   En el extremo opuesto del espectro, **Maipú** presenta el índice más bajo (**0.85**, color amarillo pálido), indicando la mayor uniformidad de todo el sector de estudio. 
*   **Pudahuel** y **Cerrillos** muestran un comportamiento casi idéntico, ambos con un índice de **0.88** (tonalidades anaranjadas claras).
*   **Interpretación analítica:** Estas tres comunas presentan una estructura socioespacial mucho más consolidada. Al tener índices de entropía bajos, se confirma matemáticamente que sus territorios están dominados por grandes bloques ininterrumpidos de las mismas tipologías habitacionales y demográficas. 

**Implicancia del Análisis Espacial:**
Este cálculo demuestra estadísticamente que los límites administrativos (las fronteras de una comuna) no sirven para entender el territorio desde una óptica comercial. El estudio señala que tratar a "Maipú" como una sola unidad geográfica podría ser viable dada su alta homogeneidad, pero intentar aplicar una regla general para "Estación Central" sería un error metodológico, ya que su extrema variabilidad interna exige un análisis micro-territorial.

## 3. Conclusión

La aplicación de técnicas de microsimulación espacial y algoritmos de clustering ha demostrado que el territorio del Sector Sur-Poniente no es un mercado uniforme. Para garantizar el éxito del nuevo canal e-commerce de la **Botillería 7LIRAS**, la expansión no debe basarse en la intuición, sino en atacar estratégicamente los nichos territoriales descubiertos. 

A partir del modelamiento de datos, se desprenden las siguientes directrices definitivas para la operación del negocio:

**Ubicación Estratégica de la Dark Store:**
El análisis territorial indica que la bodega de comercio electrónico no debe instalarse en el sector más caro, ni tampoco en el más periférico. Se recomienda ubicar la *dark store* en el corredor de transición de **Estación Central** (zonas dominadas por el clúster "Con presupusto"). Al ser la comuna con mayor Índice de Shannon (1.28), funciona como un puente geográfico: permite pagar un arriendo logístico moderado y, al mismo tiempo, garantiza tiempos de despacho en minutos hacia el nicho de los "Clientes VIP" (alta propensión de compra y altos ingresos).

**Funcionamiento del E-commerce:**
La plataforma digital no debe mostrar la misma oferta a todos los clientes. La aplicación web o las campañas de redes sociales deben segmentarse geográficamente según el clúster:
*   **Enfoque VIP y Alta Rentabilidad:** Para el sector nororiente (Estación Central límite Pudahuel), la publicidad debe priorizar destilados *premium*, licores importados y destacar "despachos express". En este grupo, la inmediatez justifica cobrar una tarifa de envío más alta, maximizando el margen por pedido.
*   **Enfoque de Rotación y "Previas":** Para las zonas de transición habitadas por jóvenes arrendatarios, el menú del e-commerce debe impulsar promociones dinámicas de fin de semana (combos de destilado + bebida + hielo). Es un cliente que busca variedad y rapidez para eventos sociales.
*   **Enfoque Familiar y Volumen:** Para las grandes extensiones homogéneas de Maipú y Cerrillos ("Los del finde" y "Caza ofertas"), la inmediatez de madrugada pierde relevancia. Aquí el negocio debe sostenerse mediante compras programadas: *packs* de cervezas de 24 unidades, cajas de vino y ofertas familiares. El incentivo principal debe ser el "despacho gratuito por compras sobre X monto", asegurando un alto volumen de ventas en cada viaje del repartidor.

En definitiva, este estudio dota a **7LIRAS** de la inteligencia comercial necesaria para focalizar su inversión. Al comprender exactamente dónde está el cliente ideal, qué presupuesto tiene y qué busca, se minimiza el riesgo financiero de la expansión digital y se maximiza la rentabilidad anual del proyecto.
