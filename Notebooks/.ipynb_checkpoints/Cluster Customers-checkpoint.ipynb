```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
```

```python
RUTA_DATOS = '../Data/Mall_Customers.csv'
SEED = 123 # Semilla ya que k-means es estocástico
```

```python
mall = pd.read_csv(RUTA_DATOS)
```

```python
mall
```

```python
variables = ['Age', 'Annual Income (k$)', 'Spending Score (1-100)']
mall_datos = mall[variables]

mall_datos.head()
```

```python
escalador = StandardScaler()
mall_escalado = escalador.fit_transform(mall_datos)
```

```python
## Cuantos cluster (k)? Utilizmaos el método del codo

wss = []
valores_k = range(1,11)
for k in valores_k:
    modelo = KMeans(n_clusters=k, n_init=25, random_state=SEED)
    modelo.fit(mall_escalado)
    wss.append(modelo.inertia_) # inertia_ es lo mismo que wss
```

```python
plt.figure(figsize=(7, 5))
plt.plot(valores_k, wss, marker="o")
plt.title("Método del codo")
plt.xlabel("Número de clusters (k)")
plt.ylabel("WSS")
plt.xticks(list(valores_k))
plt.show()
```

```python
# k obtenido a partir del método del codo
K=4

# Se ejecuta el algoritmo
kmeans = KMeans(n_clusters=K, n_init=25, random_state=SEED)
```

```python
# Se asigna un cluster a cada cliente
mall["cluster"]=kmeans.fit_predict(mall_escalado).astype(str)
```

```python
# Se ve cuantos clientes pertenecen a cada cluster
mall["cluster"].value_counts().sort_index() # Podría ser tambien un gráfico
```

## VISUALIZACIÓN DE CLUSTER

se visualiza la correlación entre varibles utilizadas 

```python
fig, ax = plt.subplots(figsize=(8, 6))
sns.scatterplot(data=mall, x="Annual Income (k$)", y="Spending Score (1-100)", 
                hue="cluster", palette="tab10", s=60, ax=ax)
ax.set_xlabel("Ingreso Anual (k$)")
ax.set_ylabel("Nivel de Gasto (1-100)")
ax.set_title("Clusters de clientes: Ingreso vs. Gasto")
plt.show()
```

```python
# Cluster 0 => alto ingreso alto gasto 
# Cluster 1 => poco ingreso poco gasto
# Cluster 2 => poco ingreso alto gasto 
# Cluster 3 => alto ingreo poco gasto 
```

```python
fig, ax = plt.subplots(figsize=(8, 6))
sns.scatterplot(data=mall, x="Age", y="Spending Score (1-100)", 
                hue="cluster", palette="tab10", s=60, ax=ax)
ax.set_xlabel("Edad")
ax.set_ylabel("Nivel de Gasto (1-100)")
ax.set_title("Clusters de clientes: Edad vs. Gasto")
plt.show()
```

```python
# Cluster 0 => alto ingreso alto gasto, adulto/joven => wnes con plata que gastan
# Cluster 1 => poco ingreso poco gasto, adulto/adulto mayor => compran poco
# Cluster 2 => poco ingreso alto gasto, joven => jovenes impulsivos
# Cluster 3 => alto ingreo poco gasto, edad 15-60 => mano wawa
```
